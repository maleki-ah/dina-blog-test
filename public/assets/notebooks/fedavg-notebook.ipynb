### FedAvg round

Each client trains locally, then the server averages the weights.

```python
clients = {"a": (120, 0.42), "b": (80, 0.51), "c": (200, 0.38)}
# (samples, local loss) per client
total = sum(n for n, _ in clients.values())
global_loss = sum(n / total * loss for n, loss in clients.values())
global_loss
```

```python
print(f"clients={len(clients)} total_samples={total}")
```
